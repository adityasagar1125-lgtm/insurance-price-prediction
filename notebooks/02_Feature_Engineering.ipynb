Binary Encoding

We do Encoding the sex and smoker in the binary form because they have only two category and Linear Regression needs the numerical data so we have to encoding but we choose binary encoding due to less feature value and also we can't drop this columns beacuse they are the important feautures for charges which is dependent on this faetures 

```python
X_train["sex"]=X_train["sex"].map({"female":1,"male":0})
X_train["smoker"]=X_train["smoker"].map({"yes":1,"no":0})
X_test["sex"]=X_test["sex"].map({"female":1,"male":0})
X_test["smoker"]=X_test["smoker"].map({"yes":1,"no":0})
X_train
```

Binary encoding converted the categorical features sex and smoker into numerical values so that they can be used by machine learning algorithms. Since both features have only two categories, binary encoding is sufficient and avoids creating unnecessary additional columns.

```python
X_train=pd.get_dummies(X_train,columns=["region"],drop_first=True,dtype=int)
X_test=pd.get_dummies(X_test,columns=["region"],drop_first=True,dtype=int)
X_train
```

Observations 
During One-Hot Encoding, one category can always be inferred from the remaining encoded columns. Keeping all dummy variables introduces perfect multicollinearity, where one feature is a linear combination of the others. This can make regression coefficients unstable and difficult to interpret. Therefore, we drop one dummy column while preserving all the information.
