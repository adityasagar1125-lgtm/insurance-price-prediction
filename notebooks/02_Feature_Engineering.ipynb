## Feature Scaling 
The numerical features in the dataset have different ranges, which may affect the performance of machine learning algorithms. To ensure that all numerical features contribute fairly to the model, we apply **Standard Scaling**. In this project, the features `age`, `bmi`, and `children` are scaled. Binary encoded features (`sex` and `smoker`) and one-hot encoded region columns are not scaled because they already contain values of 0 and 1. To prevent **data leakage**, the scaler is fitted only on the training data. The same learned parameters are then used to transform both the training and testing datasets.

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train[['age','bmi','children']]=scaler.fit_transform(
    X_train[['age','bmi','children']]
)
X_test[['age','bmi','children']]=scaler.transform(
    X_test[['age','bmi','children']]
)
```

## observations
- The numerical features (`age`, `bmi`, and `children`) have been successfully standardized. - Their values are now centered around a mean close to **0** with a standard deviation close to **1**. - Binary encoded and one-hot encoded features remain unchanged. - The scaler was fitted only on the training data, ensuring that no information from the test set was used during preprocessing. - The dataset is now fully prepared for training machine learning models.
